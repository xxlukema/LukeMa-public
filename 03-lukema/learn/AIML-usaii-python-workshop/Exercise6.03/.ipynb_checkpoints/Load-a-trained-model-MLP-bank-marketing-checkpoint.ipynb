```python
###########################
# 1. libe
#

import os
import pickle
import pandas as pd
from sklearn.neural_network import MLPClassifier
```

```python
###########################
# 2. load the model
#

path = os.getcwd() + '/bank_final_model.pkl'
with open(path, 'rb') as file:
    model = pickle.load(file)
```

```python
###########################
# 3. predict
#

feature_header = ['age','job','marital','education','default','balance','housing','loan','duration']
    
feature_data = [
    [51, 10, 1, 3, 0, 825, 0, 0, 977],
    [44, 10, 2, 2, 0, 29, 1, 0, 151]
]
    
data_frame = pd.DataFrame(feature_data, columns=feature_header)
    
pred = model.predict(data_frame)
print(pred)
```
