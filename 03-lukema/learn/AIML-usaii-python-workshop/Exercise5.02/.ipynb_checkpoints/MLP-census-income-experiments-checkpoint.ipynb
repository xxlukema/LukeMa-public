```python
################################
# 1. libs
#

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score
```

```python
################################
# 2. load data
#

data = pd.read_csv('census_income_dataset_preprocessed.csv')

################################
# Do not read feature names (dataset header)
#

# df = pd.read_csv('census_income_dataset_preprocessed', header = None)

display(data.head(2))
print(type(data))
```

```python
################################
# 2. X, Y
#

# X = data.drop('target', axis = 1)
# Y = data['target']

X = data.iloc[:, :-1]
Y = data.iloc[:, -1]

print(X.shape, Y.shape)

display(X.head(2))
print('--------------------')
display(Y.head(2))
```

```python
################################
# 3. test new dataframe
#

Z = pd.concat([X, Y], axis = 1)
Z.head(2)
```

```python
################################
# 4. split train, dev, test
#

split_factor = 0.1

X_train_tmp, X_test, Y_train_tmp, Y_test = train_test_split(X, Y, test_size = split_factor, random_state = 101)

dev_size_factor = X_test.shape[0] / X_train_tmp.shape[0]

X_train, X_dev, Y_train, Y_dev = train_test_split(X_train_tmp, Y_train_tmp, test_size = dev_size_factor, random_state = 101)

print(X_train.shape, Y_train.shape)
print(X_dev.shape, Y_dev.shape)
print(X_test.shape, Y_test.shape)

display(X_train.head(2))
```

```python
################################
# 5. define test sets
#

sets = ['Train', 'Dev', 'Test']
feature_sets = [X_train, X_dev, X_test]
target_sets = [Y_train, Y_dev, Y_test]
```

```python
################################
# 6.1. model fit/predict - experiment 1
#

model = MLPClassifier(max_iter = 500, random_state = 101)
model.fit(X_train, Y_train)

scores = {}

for idx, set in enumerate(sets):
    Y_pred = model.predict(feature_sets[idx])    
    acc_score = accuracy_score(target_sets[idx], Y_pred)
    scores[set] = round(acc_score, 4)

print(scores)
```

```python
################################
# 6.2. model fit/predict - experiment 2
#

model = MLPClassifier(max_iter = 500, random_state = 101, hidden_layer_sizes = (100, 100))
model.fit(X_train, Y_train)

scores = {}

for idx, set in enumerate(sets):
    Y_pred = model.predict(feature_sets[idx])    
    acc_score = accuracy_score(target_sets[idx], Y_pred)
    scores[set] = round(acc_score, 4)

print(scores)
```

```python
################################
# 6.3. model fit/predict - experiment 3
#

model = MLPClassifier(max_iter = 500, random_state = 101, hidden_layer_sizes = (100, 100, 100))
model.fit(X_train, Y_train)

scores = {}

for idx, set in enumerate(sets):
    Y_pred = model.predict(feature_sets[idx])    
    acc_score = accuracy_score(target_sets[idx], Y_pred)
    scores[set] = round(acc_score, 4)

print(scores)
```

```python
################################
# 6.4. model fit/predict - experiment 4
#

model = MLPClassifier(max_iter = 500, random_state = 101, hidden_layer_sizes = (150, 150))
model.fit(X_train, Y_train)

scores = {}

for idx, set in enumerate(sets):
    Y_pred = model.predict(feature_sets[idx])    
    acc_score = accuracy_score(target_sets[idx], Y_pred)
    scores[set] = round(acc_score, 4)

print(scores)
```

```python
################################
# 6.5. model fit/predict - experiment 5
#

model = MLPClassifier(max_iter = 500, random_state = 101, hidden_layer_sizes = (50, 50))
model.fit(X_train, Y_train)

scores = {}

for idx, set in enumerate(sets):
    Y_pred = model.predict(feature_sets[idx])    
    acc_score = accuracy_score(target_sets[idx], Y_pred)
    scores[set] = round(acc_score, 4)

print(scores)
```
