```python
###############################
# 1. libs
#

import pandas as pd
```

```python
###############################
# 2. load data
#

# df = pd.read_csv('census_income_dataset_preprocessed.csv', header = None)
df = pd.read_csv('census_income_dataset_preprocessed.csv')

display(df.head(2))
print(df.shape)
```

```python
###############################
# 3. Train and Test
#

X = df.iloc[:, :-1]
Y = df.iloc[:, -1]

display(X.head(2))
display(Y.head(2))

display(X.shape, Y.shape)
```

```python
###############################
# 4.1. split train, dev, test
#

from sklearn.model_selection import train_test_split

split_factor = 0.1

X_train_tmp, X_test, Y_train_tmp, Y_test = train_test_split(X, Y, test_size = split_factor, random_state = 101)

dev_split_factor = X_test.shape[0] / X_train_tmp.shape[0]

X_train, X_dev, Y_train, Y_dev = train_test_split(X_train_tmp, Y_train_tmp, test_size = dev_split_factor, random_state = 101)

print(X_train.shape, Y_train.shape)
print(X_dev.shape, Y_dev.shape)
print(X_test.shape, Y_test.shape)
```

```python
###############################
# 4.2. create a train/dev set that combines train and validation datasets
#

import numpy as np

np.random.seed(101)
    
indices_train = np.random.randint(0, len(X_train), 25)
indices_dev = np.random.randint(0, len(X_dev), 25)   
X_train_dev = pd.concat([X_train.iloc[indices_train, :], X_dev.iloc[indices_dev, :]])

# ??????????????????????????????????????????????????????
# Why this line fail? IndexingError: Too many indexers
#
# Y_train_dev = pd.concat([Y_train.iloc[indices_train, :], Y_dev.iloc[indices_dev, :]])

print(X_train_dev.shape)
display(X_train_dev.head(2))
# print(Y_train_dev.shape)
# display(Y_train_dev.head(2))
```

```python
###############################
# 4.3 prepare test sets
#

sets = ['Training', 'Validation', 'Testing']
X_sets = [X_train, X_dev, X_test]
Y_sets = [Y_train, Y_dev, Y_test]
```

```python
###############################
# 5 scores lib
#

from sklearn.metrics import accuracy_score
from sklearn.metrics import precision_score
from sklearn.metrics import recall_score
```

```python
###############################
# 6. metrics
#

metrics = {"NB":{"Acc":{},"Pre":{},"Rec":{}},
           "DT":{"Acc":{},"Pre":{},"Rec":{}},
           "SVM":{"Acc":{},"Pre":{},"Rec":{}},
           "MLP":{"Acc":{},"Pre":{},"Rec":{}}}
```

```python
###############################
# 7.1 Naive Bayes
#

from sklearn.naive_bayes import GaussianNB

model_nb = GaussianNB()
model_nb.fit(X_train, Y_train)

display(model_nb)
    
for i, set in enumerate(sets):
    pred = model_nb.predict(X_sets[i])
    score_accuracy = accuracy_score(Y_sets[i], pred)
    score_precision = precision_score(Y_sets[i], pred)
    score_recall = recall_score(Y_sets[i], pred)
    metrics['NB']['Acc'][set] = round(score_accuracy, 4)
    metrics['NB']['Pre'][set] = round(score_precision, 4)
    metrics['NB']['Rec'][set] = round(score_recall, 4)
    
print(metrics['NB'])
```

```python
###############################
# 7.2 Tree
#

from sklearn.tree import DecisionTreeClassifier

model_tree = DecisionTreeClassifier()
model_tree.fit(X, Y)

display(model_tree)

for i, set in enumerate(sets):
    pred = model_tree.predict(X_sets[i])
    score_accuracy = accuracy_score(Y_sets[i], pred)
    score_precision = precision_score(Y_sets[i], pred)
    score_recall = recall_score(Y_sets[i], pred)
    metrics['DT']['Acc'][set] = f'{score_accuracy:.4f}'
    metrics['DT']['Pre'][set] = f'{score_precision:.4f}'
    metrics['DT']['Rec'][set] = f'{score_recall:.4f}'
    
print(metrics['DT'])
```

```python
###############################
# 7.3 SVM
#

from sklearn.svm import SVC

model_svm = SVC()
model_svm.fit(X, Y)

display(model_svm)

for i, set in enumerate(sets):
    pred = model_svm.predict(X_sets[i])
    score_accuracy = accuracy_score(Y_sets[i], pred)
    score_precision = precision_score(Y_sets[i], pred)
    score_recall = recall_score(Y_sets[i], pred)
    metrics['SVM']['Acc'][set] = round(score_accuracy, 4)
    metrics['SVM']['Pre'][set] = round(score_precision, 4)
    metrics['SVM']['Rec'][set] = round(score_recall, 4)
    
print(metrics['SVM'])
```

```python
###############################
# 7.4 sklearn.neural_network MLP
#

from sklearn.neural_network import MLPClassifier

model_mlp = MLPClassifier(max_iter = 500, random_state = 101, hidden_layer_sizes = (100, 100))
model_mlp.fit(X, Y)

display(model_mlp)

for i, set in enumerate(sets):
    pred = model_mlp.predict(X_sets[i])
    score_accuracy = accuracy_score(Y_sets[i], pred)
    score_precision = precision_score(Y_sets[i], pred)
    score_recall = recall_score(Y_sets[i], pred)
    metrics['MLP']['Acc'][set] = round(score_accuracy, 4)
    metrics['MLP']['Pre'][set] = round(score_precision, 4)
    metrics['MLP']['Rec'][set] = round(score_recall, 4)
    
print(metrics['MLP'])
```

```python
###############################
# 8 print
#

for _, model_name in enumerate(metrics):
    print(model_name)
    for _, score_kind in enumerate(metrics[model_name]):
        print('   ', score_kind, metrics[model_name][score_kind])
```
