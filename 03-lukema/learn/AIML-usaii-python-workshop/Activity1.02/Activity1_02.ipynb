```python
import seaborn as sns
from sklearn.preprocessing import LabelEncoder
import numpy as np
import matplotlib.pyplot as plt
```

```python
##################################
# 1. Load dataset
#

dataset = sns.load_dataset('titanic')
dataset.head(2)
```

```python
##################################
# 2. Remove unwanted cols
#
# Features to include: sex, age, fare, class, embark_town, alone.

# drops = ['pclass', 'sibsp', 'parch', 'embarked', 'who', 'adult_male', 'deck', 'alive']
# for col in drops:
#     titanic = titanic.drop(col, axis = 1)
# titanic.head(2)
```

```python
##################################
# 3. Shallow copy dataset for desired features
#

features_desired = ['sex', 'age', 'fare', 'class', 'embark_town', 'alone']

titanic = dataset[features_desired].copy()
titanic.head(2)
```

```python
##################################
# 4. get column names
#

cols = titanic.columns.tolist()
cols
```

```python
##################################
# 5. Detect NULL columns
#
# !Trick: This must be done before transforming text fields and filling nulls
#

for col in cols:
    sum = titanic[col].isnull().sum()
    if sum > 0:
        print(col, 'nulls', sum)
```

```python
titanic.shape
```

```python
##################################
# 6. value_counts()
#

for col in cols:
    count = titanic[col].value_counts()
    print('-------', count)
```

```python
age_mean = titanic['age'].mean()
age_mean = round(age_mean)
age_mean
```

```python
##################################
# 7. Fill 'age' null with mean
#

titanic['age'] = titanic['age'].fillna(age_mean)
# titanic['age'].fillna(age_mean, inplace = True)
titanic['age'].isnull().sum()
```

```python
##################################
# 8. Transform text fields
# TODO: Detect text fields
#

features_with_text = ['sex', 'class', 'embark_town', 'alone']

enc = LabelEncoder()

for col in features_with_text:
    titanic[col] = enc.fit_transform(titanic[col].astype('str'))

titanic.head(2)
```

```python
titanic.head(2)
```

```python
##################################
# 10. Detect outliers
#

features_of_concern = ['age', 'fare']

for col in features_of_concern:
    el = titanic[col]
    min_deviation = el.mean() - 3 * el.std()    
    max_deviation = el.mean() + 3 * el.std()

    X = titanic[el >= min_deviation]
    display(X.shape)

    X = titanic[el <= max_deviation]
    display(X.shape)
```

```python
titanic.head(2)
```

```python
titanic.shape
```

```python
##################################
# 11. Normalize dataset
#

for col in cols:
    el = titanic[col]
    normalized = (el - el.std()) / (el.max() - el.min())
    titanic[col] = normalized
```
