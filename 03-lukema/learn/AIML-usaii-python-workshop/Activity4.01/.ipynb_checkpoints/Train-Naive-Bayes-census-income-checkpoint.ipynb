```python
################################
# 1. libs
#

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
```

```python
################################
# 2. load data
#

data = pd.read_csv('census_income_dataset_preprocessed.csv')

################################
# Do not read feature names (dataset header)
#

# df = pd.read_csv('census_income_dataset_preprocessed', header = None)

print(data.head(2))
print(type(data))
```

```python
################################
# 2. load data
#

X = data.drop('target', axis = 1)
Y = data['target']

print(X.shape, Y.shape)

print(X.head(2))
print('--------------------')
print(Y.head(2))
```

```python
################################
# 3. new dataframe
#

Z = pd.concat([X, Y], axis = 1)
Z.head(2)
```

```python
################################
# 4. split train, dev, test
#

split_factor = 0.1

X_train_tmp, X_test, Y_train_tmp, Y_test = train_test_split(X, Y, test_size = split_factor, random_state = 101)

dev_size_factor = X_test.shape[0] / X_train_tmp.shape[0]

X_train, X_dev, Y_train, Y_dev = train_test_split(X_train_tmp, Y_train_tmp, test_size = dev_size_factor, random_state = 101)

print(X_train.shape, Y_train.shape)
print(X_dev.shape, Y_dev.shape)
print(X_test.shape, Y_test.shape)
```

```python
################################
# 5. model
#

model = GaussianNB()

################################
# UserWarning: X does not have valid feature names, but GaussianNB was fitted with feature names
#

# model.fit(X_train, Y_train)    # Do not train model with feature names

################################
# Do not train model with feature names
#

model.fit(X_train.values, Y_train.values)
```

```python
################################
# 6. predict
#

pred = model.predict([[39, 6, 13, 4, 0, 2174, 0, 40, 38]])
print(pred)
```
