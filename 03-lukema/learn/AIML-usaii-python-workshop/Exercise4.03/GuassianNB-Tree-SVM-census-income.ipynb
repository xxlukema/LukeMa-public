```python
###############################
# 1. libs
#

import pandas as pd
```

```python
###############################
# 2. load data
#

# df = pd.read_csv('census_income_dataset_preprocessed.csv', header = None)
df = pd.read_csv('census_income_dataset_preprocessed.csv')

display(df.head(2))
```

```python
###############################
# 3. split
#

X = df.iloc[:, :-1]
Y = df.iloc[:, -1]

display(X.head(2))
display(Y.head(2))

display(X.shape, Y.shape)
```

```python
###############################
# 4.0 test data
#

test = [[39, 6, 13, 4, 0, 2174, 0, 40, 38]]
```

```python
###############################
# 4.1 Naive Bayes
#

from sklearn.naive_bayes import GaussianNB

model_nb = GaussianNB()
model_nb.fit(X.values, Y.values)

display(model_nb)

pred_nb = model_nb.predict(test)
print(pred_nb)
```

```python
###############################
# 4.2 Tree
#

from sklearn.tree import DecisionTreeClassifier

model_tree = DecisionTreeClassifier()
model_tree.fit(X.values, Y.values)

display(model_tree)

pred_tree = model_tree.predict(test)
print(pred_tree)
```

```python
###############################
# 4.3 SVM
#

from sklearn.svm import SVC

model_svm = SVC()
model_svm.fit(X.values, Y.values)

display(model_svm)

pred_svm = model_svm.predict(test)
print(pred_svm)
```
