```python
###############################
# 1. libs
#

import pandas as pd
from sklearn.svm import SVC
```

```python
###############################
# 2. load data
#

# df = pd.read_csv('census_income_dataset_preprocessed.csv', header = None)
df = pd.read_csv('census_income_dataset_preprocessed.csv')

display(df.head(2))
```

```python
###############################
# 3. split
#

X = df.iloc[:, :-1]
Y = df.iloc[:, -1]

display(X.head(2))
display(Y.head(2))

display(X.shape, Y.shape)
```

```python
###############################
# 4. SVM model
#

model = SVC()
model.fit(X.values, Y.values)
```

```python
###############################
# 5. predict
#

pred = model.predict([[39, 6, 13, 4, 0, 2174, 0, 40, 38]])
print(pred)
```
