```python
##########################
# split-wine
#

from sklearn.datasets import load_wine
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
##########################
# load data
# The output from the preceding function is a dictionary-like object, which separates the features
# (callable as data) from the target (callable as target) into two attributes.
#

data = load_wine()
type(data)
```

```python
##########################
# Convert data into pandas DataFrames
#

X = pd.DataFrame(data.data)
Y = pd.DataFrame(data.target)

print(X.shape, Y.shape)
print(X.head(1))
print(Y.head(1))
```

```python
##########################
# split data (small dataset: 60/20/20%, large dataset: 98/1/1%)
#

X, X_test, Y, Y_test = train_test_split(X, Y, test_size = 0.2)
```

```python
##########################
# train/validation split
#

print(X.shape, X_test.shape, Y.shape, Y_test.shape)
```

```python
##########################
# Step 1/2: train/test split
#

dev_size = X_test.shape[0]/X.shape[0]
print(dev_size)
```

```python
##########################
# Step 2/2: train/test split
#

X_train, X_dev, Y_train, Y_dev = train_test_split(X, Y, test_size = dev_size)
print(X_train.shape, Y_train.shape)
print(X_dev.shape, Y_dev.shape)
print(X_test.shape, Y_test.shape)
```
