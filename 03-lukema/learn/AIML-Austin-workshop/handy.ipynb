```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
data = {'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
        'Age': [25, 30, 35, 28, 32],
        'City': ['New York', 'London', 'Paris', 'Tokyo', 'Sydney']}

df = pd.DataFrame(data)
df
```

```python

# Filter rows where Age is greater than 30
filtered_df = df[df['Age'] > 30]
print(filtered_df)
# Output:
#    Name  Age    City
# 2 Charlie   35   Paris
# 4     Eve   32  Sydney
```

```python

# Filter rows where City is 'London'
filtered_df = df[df['City'] == 'London']
print(filtered_df)
# Output:
#   Name  Age    City
# 1  Bob   30  London
```

```python

# Filter rows where Age is less than or equal to 28
filtered_df = df[df['Age'] <= 28]
print(filtered_df)
# Output:
#    Name  Age      City
# 0  Alice   25  New York
# 3  David   28     Tokyo
```

```python

# Filter rows based on multiple conditions using logical operators
filtered_df = df[(df['Age'] > 25) & (df['City'] != 'Tokyo')]
print(filtered_df)
# Output:
#    Name  Age      City
# 1    Bob   30    London
# 2  Charlie   35     Paris
# 4    Eve   32    Sydney
```

```python

# Filter rows where a column value is in a list of values
cities = ['London', 'Sydney']
filtered_df = df[df['City'].isin(cities)]
print(filtered_df)
# Output:
#   Name  Age    City
# 1  Bob   30  London
# 4  Eve   32  Sydney
```
