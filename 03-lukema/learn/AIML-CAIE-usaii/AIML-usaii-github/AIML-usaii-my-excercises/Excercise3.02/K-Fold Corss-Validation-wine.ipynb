```python
########################
# K-Fold Cross-Validation
#

from sklearn.datasets import load_wine
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.model_selection import KFold
```

```python
########################
# load_wine and convert to pandas.DataFrame
#

data = load_wine()
X = pd.DataFrame(data.data)
Y = pd.DataFrame(data.target)
```

```python
########################
# Step 1/2. 60/20/20% split
#

X_train_tmp, X_test, Y_train_tmp, Y_test = train_test_split(X, Y, test_size = 0.2)
```

```python
########################
# Step 2/2. 60/20/20% split
#

split_factor = X_test.shape[0]/X_train_tmp.shape[0]

X_train, X_tune, Y_train, Y_tune = train_test_split(X_train_tmp, Y_train_tmp, test_size = split_factor)
```

```python
########################
# Train here
#

print(X_train.shape, Y_train.shape)
print(X_tune.shape, Y_tune.shape)
print(X_test.shape, Y_test.shape)
```

```python
########################
# K-Fold
#

K = 10

kf = KFold(n_splits = K)
```

```python
########################
# K-Fold: split indexes
#

splits = kf.split(X)
```

```python
##################################
# Use 10% as test instances
# Create X_train_tmp, Y_train_tmp for further processing
#

test_factor = 0.1

X_train_tmp, X_test, Y_train_tmp, Y_test = train_test_split(X, Y, test_size = test_factor)
```

```python
##################################
# KFold: 10-Fold
# Further splt X_train_tmp, Y_train_tmp
#

splits = kf.split(X_train_tmp)
```
