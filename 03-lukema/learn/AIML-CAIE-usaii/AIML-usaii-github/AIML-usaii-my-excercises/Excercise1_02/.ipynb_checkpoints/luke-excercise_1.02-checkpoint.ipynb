```python
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt

dataset = sns.load_dataset('tips')
display(dataset.head(3))
```

```python
size = dataset['size']
size.loc[:12] = np.nan
size.head(3)
```

```python
size.shape
```

```python
size.isnull().sum()
```

```python
mean = size.mean()
mean = round(mean)
mean
```

```python
size.fillna(mean, inplace=True)
size.head(3)
```

```python
plt.hist(size)
plt.show()
```

```python
min_val = size.mean() - 3 * size.std()
min_val
```

```python
max_val = size.mean() + 3 * size.std()
max_val
```

```python
outliers = size[size > max_val]
outliers.count()
```

```python
print(outliers)
```

```python
size.shape
```

```python
size = size[size <= max_val]
size.shape
```
