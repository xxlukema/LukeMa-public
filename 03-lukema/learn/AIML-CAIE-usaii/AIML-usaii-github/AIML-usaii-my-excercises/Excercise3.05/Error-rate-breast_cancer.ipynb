```python
########################
# libs
#

from sklearn.datasets import load_breast_cancer
import pandas as pd
from sklearn.model_selection import train_test_split
import numpy as np
from sklearn import tree
from sklearn.metrics import recall_score
```

```python
########################
# load data
#

data = load_breast_cancer()
X = pd.DataFrame(data.data)
Y = pd.DataFrame(data.target)
```

```python
########################
# split
#

split_factor = 0.1
X_train_tmp, X_test, Y_train_tmp, Y_test = train_test_split(X, Y, test_size = split_factor, random_state = 101)

dev_train_ratio = X_test.shape[0] / X_train_tmp.shape[0]
X_train, X_dev, Y_train, Y_dev = train_test_split(X_train_tmp, Y_train_tmp, test_size = dev_train_ratio, random_state = 101)

print(X_train.shape, X_dev.shape, X_test.shape)
print(Y_train.shape, Y_dev.shape, Y_test.shape)
```

```python
########################
# create a train/dev set that combines train and validation datasets
#

np.random.seed(101)

indices_train = np.random.randint(0, len(X_train), 25)
indices_dev = np.random.randint(0, len(X_dev), 25)

X_train_dev = pd.concat([X_train.iloc[indices_train, :], X_dev.iloc[indices_dev, :]])
Y_train_dev = pd.concat([Y_train.iloc[indices_train, :], Y_dev.iloc[indices_dev, :]])

print(X_train_dev.shape, Y_train_dev.shape)
```

```python
########################
# tree, fit
#

model = tree.DecisionTreeClassifier(random_state = 101)
est_model = model.fit(X_train, Y_train)
```

```python
########################
# predict, recall_score using all kind of sets
#

sets = ['Training', 'Train/dev', 'Validation', 'Testing']
X_sets = [X_train, X_train_dev, X_dev, X_test]
Y_sets = [Y_train, Y_train_dev, Y_dev, Y_test]

scores = {}

for i, set in enumerate(sets):
    pred = est_model.predict(X_sets[i])
    score = recall_score(Y_sets[i], pred)
    scores[set] = score

print(scores)
```

```python
########################
# Errors
#

errors = {}

errors['Bayes error'] = 0.0

for score in scores:
    # errors[score] = float('{:.4f}'.format(1 - scores[score]))
    # or
    errors[score] = round(1 - scores[score], 4)

print(errors)
```
