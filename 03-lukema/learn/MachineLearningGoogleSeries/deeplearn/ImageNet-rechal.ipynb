Deep Learn ImageNet Using Tensorflow

```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
from tqdm import tqdm
```

```python
# 1. Load image using tensorflow

from tensorflow.keras.preprocessing import image

# filename = 'dancers.jpg'
filename = 'rechal.jpg'

img = image.load_img(filename, target_size=(224, 224))
img
```

```python
# 2. Preprocess image

resized_image = image.img_to_array(img)
final_image = np.expand_dims(resized_image, axis=0)
# final_image = tf.keras.applications.mobilenet.preprocess_input(final_image)
final_image = tf.keras.applications.mobilenet_v2.preprocess_input(final_image)

print(resized_image.shape)
print(final_image.shape)
```

```python
# 3. Find a model and predict

model = tf.keras.applications.mobilenet.MobileNet()
# model = tf.keras.applications.mobilenet_v2.MobileNetV2()
# model = tf.keras.applications.MobileNetV2()
# model = tf.keras.applications.DenseNet201()

prediction = model.predict(final_image)
```

```python
from tensorflow.keras.applications import imagenet_utils
result = imagenet_utils.decode_predictions(prediction)

plt.imshow(img)
```
