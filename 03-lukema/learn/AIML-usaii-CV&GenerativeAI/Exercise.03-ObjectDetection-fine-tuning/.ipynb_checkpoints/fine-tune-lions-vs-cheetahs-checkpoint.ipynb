```python
import json
import math
import keras_cv
import numpy as np
import tensorflow as tf
import tensorflow_datasets as tfds
import keras
from keras import losses
from keras import metrics
import matplotlib.pyplot as plt
```

```python
# !wget -nc https://archive.org/download/lions_cheetahs/lions_cheetahs.zip
# !unzip -nq lions_cheetahs.zip -d lions_cheetahs
# !ls lions_cheetahs
```

```python
train_ds, val_ds = tf.keras.utils.image_dataset_from_directory(
    "lions_cheetahs/images",
    image_size=(224, 224),
    validation_split=0.3,
    subset="both",
    seed=123
)
```

```python
train_ds
```

```python
val_ds
```

```python
val_ds.class_names
```

```python
train_ds = train_ds.shuffle(10 * 32, reshuffle_each_iteration=True)
```

```python
images = next(iter(train_ds.take(1)))[0]

print(images.shape)

keras_cv.visualization.plot_image_gallery(images=images, value_range=(0, 255))
```

```python
model = keras_cv.models.ImageClassifier.from_preset(
    "efficientnetv2_b0_imagenet",
    num_classes=2
)

model.compile(
    loss = "sparse_categorical_crossentropy",
    optimizer = tf.keras.optimizers.AdamW(),
    metrics = ["accuracy"]
)
```

```python
results = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=10
)
```

```python
plt.plot(results.history["loss"], label="loss")
plt.plot(results.history["val_loss"], label="val_loss")
plt.legend()
```

```python
plt.plot(results.history["accuracy"], label="accuracy")
plt.plot(results.history["val_accuracy"], label="val_accuracy")
plt.legend()
```
