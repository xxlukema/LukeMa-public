```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
rng = np.random.default_rng(0)  # Create a random generator with a seed

# Generate synthetic data
n_samples = 1000

X = rng.standard_normal(n_samples)  # Generate random numbers using the generator
display(X[:5])

plt.figure(figsize=(10, 6))
plt.plot(X, marker='.', linestyle='-', markersize=1, label='Random Numbers')
plt.title('Random Numbers Plot')
plt.xlabel('Sample Index')
plt.ylabel('Value')
plt.legend()
plt.grid()
plt.show()

plt.hist(X, bins=30, density=True, alpha=0.5, color='g')
plt.title('Histogram of Random Numbers')
plt.xlabel('Value')
plt.ylabel('Density')
plt.show()
```

```python
plt.scatter(X, X, alpha=0.5)
plt.title('Scatter Plot of Random Numbers vs. Squared Values')
plt.xlabel('Random Numbers')
plt.ylabel('Squared Values')
plt.grid()
plt.show()
```

```python
Y = X.cumsum()  # Cumulative sum of the random numbers
plt.plot(Y, marker='.', linestyle='-', markersize=1, label='Cumulative Sum')
plt.title('Cumulative Sum of Random Numbers')
plt.xlabel('Sample Index')
plt.ylabel('Cumulative Sum')
plt.legend()
plt.grid()
plt.show()
```

```python
X = rng.multivariate_normal([0, 0], [[1, -0.5], [-0.5, 2]], size=1000)

plt.scatter(X[:, 0], X[:, 1], alpha=0.5)
plt.title('Scatter Plot of Multivariate Normal Distribution')
plt.xlabel('X1')
plt.ylabel('X2')
plt.grid()
plt.show()

print(X[:5])  # Display the first 5 samples of the multivariate normal distribution
```

```python
df = pd.DataFrame(X, columns=['X1', 'X2'])
print(df['X1'].mean())
# print(df['X1'].cov())  # Covariance of X1

```
